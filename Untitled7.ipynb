## 1. Problema de los Gradientes Inestables e Inicialización

Como mencionas, los *gradientes desvanecientes* y *explosivos* son un problema fundamental en el entrenamiento de redes neuronales profundas. Una **inicialización de pesos** adecuada es crucial para asegurar que los gradientes se mantengan en un rango razonable, permitiendo que la red aprenda de forma estable.

### Inicialización de Glorot (Xavier) y He

*   **Glorot (Xavier) Initialization:** Diseñada para mantener la varianza de las salidas y los gradientes igual a la de las entradas en cada capa. Es adecuada para funciones de activación con propiedades lineales alrededor de cero, como la `tanh` o la `sigmoid`.
*   **He Initialization:** Una variante de Glorot, optimizada para funciones de activación no saturantes como `ReLU` y sus variantes. Asegura que la varianza de las activaciones no cambie drásticamente a medida que la información fluye a través de la red.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# Definir una semilla para reproducibilidad
tf.random.set_seed(42)
np.random.seed(42)
```

A continuación, crearemos una red neuronal simple y mostraremos cómo se aplican las diferentes estrategias de inicialización en TensorFlow/Keras. Observa los valores de los pesos inicializados para entender la diferencia.

```python
# Función para crear y mostrar un modelo con una inicialización específica
def create_and_display_model(initializer, activation):
    model = models.Sequential([
        layers.Input(shape=(784,)), # Ejemplo con 784 características de entrada
        layers.Dense(256, activation=activation, kernel_initializer=initializer, name=f'hidden_layer_1_{initializer}'),
        layers.Dense(128, activation=activation, kernel_initializer=initializer, name=f'hidden_layer_2_{initializer}'),
        layers.Dense(10, activation='softmax', name=f'output_layer_{initializer}')
    ])
    print(f"\n--- Modelo con inicialización: {initializer} y activación: {activation} ---")
    model.summary()

    # Mostrar los pesos de la primera capa oculta
    print(f"\nPesos de la primera capa oculta (kernel): {model.layers[0].name}")
    weights = model.layers[0].get_weights()[0]
    print(f"Forma de los pesos: {weights.shape}")
    print(f"Varianza de los pesos: {np.var(weights):.4f}")
    print(f"Media de los pesos: {np.mean(weights):.4f}")

    return model

# Modelo con inicialización Glorot (Xavier) y activación 'tanh'
model_glorot_tanh = create_and_display_model('glorot_uniform', 'tanh')

# Modelo con inicialización He y activación 'relu'
model_he_relu = create_and_display_model('he_normal', 'relu')
```

Como puedes ver, los valores iniciales de los pesos varían dependiendo del inicializador elegido, lo cual influye en la estabilidad del entrenamiento. Esta es solo una de las muchas técnicas que mencionaste en tus notas.
